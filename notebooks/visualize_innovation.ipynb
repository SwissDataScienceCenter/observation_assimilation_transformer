```python
import torch
import matplotlib.pyplot as plt
import numpy as np
```

```python
from oat.model_module import distributions
```

```python
dist = distributions.LogNormalOAT(loc=torch.tensor(2.0), scale=torch.tensor(1.0))
```

```python
xmin, xmax = 1e-3, 50
x = torch.linspace(xmin, xmax, steps=1000)

PDF_COLOR = "#1f77b4"
INNOV_COLOR = "#d62728"

fig, ax1 = plt.subplots(figsize=(7, 4), dpi=150)
fig.patch.set_facecolor("white")
ax1.set_facecolor("#f8f8f8")
ax1.grid(True, color="white", linewidth=0.8)
ax1.set_axisbelow(True)

ax1.plot(x.numpy(), torch.exp(dist.log_prob(x)).detach().numpy(),
         color=PDF_COLOR, linewidth=2, label="PDF")
ax1.set_xlabel("x")
ax1.set_ylabel("PDF", color=PDF_COLOR)
ax1.tick_params(axis="y", colors=PDF_COLOR)
ax1.spines["left"].set_color(PDF_COLOR)
ax1.spines["top"].set_visible(False)

ax2 = ax1.twinx()
ax2.plot(x.numpy(), dist.innovation(x).detach().numpy(),
         color=INNOV_COLOR, linewidth=2, label="Innovation")
ax2.set_ylabel("Innovation", color=INNOV_COLOR)
ax2.tick_params(axis="y", colors=INNOV_COLOR)
ax2.spines["right"].set_color(INNOV_COLOR)
ax2.spines["top"].set_visible(False)

handles1, labels1 = ax1.get_legend_handles_labels()
handles2, labels2 = ax2.get_legend_handles_labels()
fig.legend(handles1 + handles2, labels1 + labels2,
           loc="upper center", ncol=2, framealpha=0.9,
           bbox_to_anchor=(0.5, 1.0))

ax1.set_title("LogNormal", fontsize=13, fontweight="bold", pad=10)
ax1.set_xlim(xmin, xmax)
plt.tight_layout(rect=[0, 0, 1, 0.92])
plt.show()
```
