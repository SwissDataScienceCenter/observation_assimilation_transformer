```python
import torch
import matplotlib.pyplot as plt
import numpy as np
```

```python
from oat.model_module import distributions
```

```python
dist = distributions.LogNormalOAT(loc=torch.tensor(2.0), scale=torch.tensor(1.0))
```

```python
xmin, xmax = 1e-3, 50
x = torch.linspace(xmin, xmax, steps=1000)

fig, ax1 = plt.subplots()


ax1.plot(x.numpy(), torch.exp(dist.log_prob(x)).detach().numpy(), color="blue", label="PDF")
ax1.set_ylabel("Empirical PDF")

# Second axis
ax2 = ax1.twinx()


y = dist.innovation(x)

ax2.plot(x.numpy(), y.detach().numpy(), color="red", label="Innovation")
ax2.set_ylabel("Innovation")

# Combine legends
handles1, labels1 = ax1.get_legend_handles_labels()
handles2, labels2 = ax2.get_legend_handles_labels()

# Place legend outside (right side)
ax1.legend(
    handles1 + handles2,
    labels1 + labels2,
    loc="center left",
    bbox_to_anchor=(1.15, 0.5),
)

# Make room for the legend
# plt.tight_layout()
plt.title('ZeroInflatedLogNormal')
ax1.set_xlim(-1, 50)
plt.show()
```
